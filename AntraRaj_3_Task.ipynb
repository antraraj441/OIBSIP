```python
import pandas as pd
import numpy as np

df = pd.read_csv("messy_customer_data.csv")

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
original_df = df.copy()
```

Check basic information

```python
print("Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

Data Quality Report

```python
quality_report = pd.DataFrame({
    "Column": df.columns,
    "Null_Count": df.isnull().sum().values,
    "Data_Type": df.dtypes.astype(str).values,
    "Unique_Values": df.nunique().values
})

quality_report
```

```python
df.describe()
```

Check Duplicate Rows

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

```

```python
df[df.duplicated()]
```

```python
df = df.drop_duplicates()

print("Duplicates after cleaning:", df.duplicated().sum())
```

Missing Data Analysis

```python
missing = df.isnull().sum()

print(missing)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": missing_percentage
})

missing_report
```

Handle Missing Values

```python
# Fill missing Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Salary with median
df["Salary"] = pd.to_numeric(df["Salary"], errors="coerce")
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Fill missing City with mode
df["City"] = df["City"].fillna(df["City"].mode()[0])

# Fill missing Rating with median
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
df["Rating"] = df["Rating"].fillna(df["Rating"].median())

# Check missing values
print(df.isnull().sum())
```

Standardise Inconsistent Formatting

```python
# Standardise Gender
df["Gender"] = df["Gender"].astype(str).str.strip().str.lower()

df["Gender"] = df["Gender"].replace({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female"
})

print(df["Gender"].value_counts())
```

```python
df["City"] = df["City"].astype(str).str.strip().str.title()

print(df["City"].value_counts())
```

```python
df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    errors="coerce"
)
```

```python
print(df["Join_Date"].head())
```

Data Type Correction

```python
print(df.dtypes)
```

```python
df["Customer_ID"] = df["Customer_ID"].astype("string")
```

```python
df["Age"] = df["Age"].astype(float)
df["Salary"] = df["Salary"].astype(float)
df["Rating"] = df["Rating"].astype(float)
```

```python
df["Gender"] = df["Gender"].astype("category")
df["City"] = df["City"].astype("category")
```

```python
print(df.dtypes)
```

Detect Invalid Age values

```python
invalid_age = df[(df["Age"] < 0) | (df["Age"] > 100)]

print(invalid_age)
```

```python
df.loc[(df["Age"] < 0) | (df["Age"] > 100), "Age"] = np.nan
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
print(df["Age"].describe())
```

Outlier Detection Using IQR

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
salary_outliers = df[
    (df["Salary"] < lower_limit) |
    (df["Salary"] > upper_limit)
]

print("Number of Salary Outliers:", len(salary_outliers))

salary_outliers
```

Detect Rating Anomaly

```python
invalid_rating = df[
    (df["Rating"] < 1) |
    (df["Rating"] > 5)
]

print(invalid_rating)
```

```python
df.loc[
    (df["Rating"] < 1) | (df["Rating"] > 5),
    "Rating"
] = np.nan
```

```python
df["Rating"] = df["Rating"].fillna(df["Rating"].median())
```

```python
print(df["Rating"].describe())
```
